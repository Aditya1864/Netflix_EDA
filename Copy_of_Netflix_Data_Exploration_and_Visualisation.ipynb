##Data Preprocessing

```python
df_clean = df.copy()
```

```python
df_clean["date_added"] = pd.to_datetime(
    df_clean["date_added"],
    errors="coerce"
)
```

```python
df_clean.info()
```

```python
mask = df_clean["rating"].str.contains("min", na=False)

df_clean.loc[mask, "duration"] = df_clean.loc[mask, "rating"]
df_clean.loc[mask, "rating"] = np.nan
```

```python
df_clean["director"] = df_clean["director"].fillna("Unknown")
df_clean["cast"] = df_clean["cast"].fillna("Unknown")
df_clean["country"] = df_clean["country"].fillna("Unknown")
df_clean["rating"] = df_clean["rating"].fillna("Not Rated")
```

###Unnesting

```python
country_df = df_clean.copy()

country_df["country"] = country_df["country"].str.split(",")

country_df = country_df.explode("country")

country_df["country"] = country_df["country"].str.strip()
```

```python
director_df = df_clean.copy()

director_df["director"] = director_df["director"].str.split(",")

director_df = director_df.explode("director")

director_df["director"] = director_df["director"].str.strip()
```

```python
cast_df = df_clean.copy()

cast_df["cast"] = cast_df["cast"].str.split(",")

cast_df = cast_df.explode("cast")

cast_df["cast"] = cast_df["cast"].str.strip()
```

```python
genre_df = df_clean.copy()

genre_df["listed_in"] = genre_df["listed_in"].str.split(",")

genre_df = genre_df.explode("listed_in")

genre_df["listed_in"] = genre_df["listed_in"].str.strip()
```

###Create New Date features

```python
df_clean["year_added"] = df_clean["date_added"].dt.year

df_clean["month_added"] = df_clean["date_added"].dt.month_name()

df_clean["day_added"] = df_clean["date_added"].dt.day_name()
```

###Convert to categories

```python
df_clean["type"] = df_clean["type"].astype("category")
df_clean["rating"] = df_clean["rating"].astype("category")
```

##Visual Analysis

```python
plt.figure(figsize=(6,4))
sns.countplot(data=df_clean, x="type")
plt.title("Distribution of Movies and TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Count")
plt.show()
```

```python
rating_counts = df_clean["rating"].value_counts()

plt.figure(figsize=(8,6))
sns.barplot(
    x=rating_counts.values,
    y=rating_counts.index
)

plt.title("Distribution of Content Ratings")
plt.xlabel("Count")
plt.ylabel("Rating")
plt.show()
```

```python
plt.figure(figsize=(12,5))
sns.histplot(df_clean["release_year"], bins=30)

plt.title("Distribution of Release Years")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```

```python
year_added = (
    df_clean["year_added"]
    .value_counts()
    .sort_index()
)

plt.figure(figsize=(12,5))

sns.lineplot(
    x=year_added.index,
    y=year_added.values,
    marker="o"
)

plt.title("Titles Added to Netflix by Year")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")

plt.show()
```

```python
top_country = country_df[country_df["country"] != 'Unknown']['country'].value_counts().head(10)

plt.figure(figsize=(10,5))

sns.barplot(
    x=top_country.values,
    y=top_country.index
)

plt.title("Top 10 Content Producing Countries")
plt.xlabel("Titles")
plt.ylabel("Country")

plt.show()
```

```python
top_genre = genre_df[genre_df["listed_in"] != "Unknown"]["listed_in"].value_counts().head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_genre.values,
    y=top_genre.index
)

plt.title("Top 10 Genres")
plt.xlabel("Titles")
plt.ylabel("Genre")

plt.show()
```

```python
top_directors = (
    director_df[director_df["director"] != "Unknown"]["director"]
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10,6))
sns.barplot(x=top_directors.values, y=top_directors.index)

plt.title("Top 10 Directors on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Director")

plt.show()
```

```python
top_actors = (
    cast_df[cast_df["cast"] != "Unknown"]["cast"]
    .value_counts()
    .head(10)
)

plt.figure(figsize=(10,6))
sns.barplot(x=top_actors.values, y=top_actors.index)

plt.title("Top 10 Actors on Netflix")
plt.xlabel("Number of Titles")
plt.ylabel("Actor")

plt.show()
```

```python
plt.figure(figsize=(10,6))

sns.countplot(
    data=df_clean,
    y="rating",
    hue="type",
    order=df_clean["rating"].value_counts().index
)

plt.title("Content Rating by Content Type")
plt.xlabel("Number of Titles")
plt.ylabel("Rating")

plt.show()
```

```python
year_type = (
    df_clean.groupby(["year_added", "type"])
    .size()
    .reset_index(name="count")
)

plt.figure(figsize=(12,6))

sns.lineplot(
    data=year_type,
    x="year_added",
    y="count",
    hue="type",
    marker="o"
)

plt.title("Content Added to Netflix by Year and Type")
plt.xlabel("Year Added")
plt.ylabel("Number of Titles")

plt.show()
```

```python
top10 = (
    country_df[country_df["country"] != "Unknown"]["country"]
    .value_counts()
    .head(10)
    .index
)

country_top = country_df[country_df["country"].isin(top10)]

plt.figure(figsize=(12,6))

sns.countplot(
    data=country_top,
    y="country",
    hue="type",
    order=top10
)

plt.title("Top Producing Countries by Content Type")
plt.xlabel("Number of Titles")
plt.ylabel("Country")

plt.show()
```

```python
month_order = [
    "January","February","March","April","May","June",
    "July","August","September","October","November","December"
]

heatmap_data = pd.crosstab(
    df_clean["year_added"],
    df_clean["month_added"]
)

heatmap_data = heatmap_data.reindex(columns=month_order)

plt.figure(figsize=(12,8))

sns.heatmap(
    heatmap_data,
    cmap="YlGnBu",
    annot=False
)

plt.title("Titles Added by Month and Year")
plt.xlabel("Month")
plt.ylabel("Year")

plt.show()
```
